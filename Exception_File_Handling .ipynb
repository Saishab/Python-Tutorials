#### Exception Handling 

```python
x = 10
y =  0

z = x / y 
print(z)
```

```python
# try and except 
try:
    a = 10
    b = 0
    print(a / b)
except:
    print("Cannot divide by zero")
```

```python
# else 

try:
    x = int(input("Enter a number: "))
    result = 10 / x
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result is:", result)
```

```python
 ## raise an exception 

age = int(input("Enter age: "))

if age < 18:
    #print("You can't vote")
    raise Exception("Umer badayera aaija")
else:
    print("You can vote")
```
