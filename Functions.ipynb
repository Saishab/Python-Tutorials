#### Recursion 
Recursion is when a function calls itself. It must have a base condition to stop the recursion.

```python
def countdown(n):
    if n == 0:
        print("Completed!")
    else:
        print(n)
        countdown(n-1)
```

```python
countdown(10)
```

```python
def factorial(n):
    if n == 0 or n == 1:   # this is base case
        return 1
    else:
        return n * factorial(n-1)   # this is recursive call
```
